```python
# Combine first two rows into a single header
new_columns = []

for col_name, fy in zip(df1.iloc[3], df1.iloc[4]):
    if pd.isna(fy):
        new_columns.append(str(col_name).strip())
    else:
        new_columns.append(f"{col_name}_{fy}".strip())

df1.columns = new_columns

new_columns = []

for col_name, fy in zip(df2.iloc[3], df2.iloc[4]):
    if pd.isna(fy):
        new_columns.append(str(col_name).strip())
    else:
        new_columns.append(f"{col_name}_{fy}".strip())

df2.columns = new_columns
```

```python
df1 = df1.drop(df1.index[:6]).reset_index(drop = True)
df2 = df2.drop(df2.index[:6]).reset_index(drop = True)
```

```python
df1 = df1.loc[:, ~df1.columns.duplicated()]
df1.head()
```

```python
df_merged = pd.merge(df1, df2, on = "SP_ENTITY_ID", how = "inner", validate = "one_to_one")
df_merged = df_merged.drop(columns = ["SP_ENTITY_NAME_x"])
df_merged.head()
```

```python
import re
id_cols = ["SP_ENTITY_ID", "SP_COMPANY_NAME", "SP_TICKER"]
value_cols = [
    c for c in df_merged.columns
    if re.search(r"_FY\d{4}(_[xy])?$", c)]
```

#MELTING THE DATA 

```python
df_long = df_merged.melt(id_vars = id_cols, value_vars = value_cols, var_name = "variable", value_name = "value")
df_long["fiscal_year"] = ( df_long["variable"].str.extract(r"_fy(\d{4})", flags = re.IGNORECASE))
df_long["fiscal_year"] = df_long["fiscal_year"].astype("Int64")
```

```python
df_long["base_variable"] = (df_long["variable"].str.lower()
                            .str.replace(r"_fy\d{4}", "", regex=True)
                            .str.replace(r"_[xy]$", "", regex=True))
df_long[["base_variable","fiscal_year"]].head()
```

```python
df_panel = df_long.pivot_table(
    index = id_cols + ["fiscal_year"],
    columns = "base_variable",
    values = "value",
    aggfunc = "first").reset_index()
df_panel.columns.name = None
df_panel = df_panel.sort_values(
    ["SP_ENTITY_ID", "fiscal_year"]
).reset_index(drop=True)

df_panel.head()                            
```
